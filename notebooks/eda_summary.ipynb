# AgriShield Crop and Pest EDA

Exploratory analysis of the two Kaggle sources after median imputation, unit standardization, and SMOTE class balancing.

```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")
csv_path = Path("../data/processed/processed_crop_pest.csv")
df = pd.read_csv(csv_path)
print(df.shape)
df.head()
```

```python
print("Dataset sources:")
print(df["source"].value_counts())
print("\nMissing values:")
print(df.isna().sum()[df.isna().sum() > 0])
print("\nNumeric summary:")
df.describe(include="number").T
```

```python
source_counts = df.groupby(["source", "label"]).size().rename("count").reset_index()
plt.figure(figsize=(14, 7))
sns.barplot(data=source_counts, x="label", y="count", hue="source")
plt.xticks(rotation=65, ha="right")
plt.title("Class Distribution after SMOTE")
plt.tight_layout()
plt.show()
```

```python
numeric = df.select_dtypes(include="number").columns
df[numeric].hist(figsize=(16, 12), bins=30, edgecolor="black")
plt.suptitle("Distributions of Standardized Numeric Features")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(12, 8))
sns.heatmap(df[numeric].corr(), annot=True, fmt=".2f", cmap="coolwarm", center=0)
plt.title("Numeric Feature Correlations")
plt.tight_layout()
plt.show()
```

## Data preparation summary

- Crop and pest categorical labels are preserved in `label`; source-specific labels are in `crop_label` and `pest_name`.
- Numeric missing values were median-imputed.
- Temperature is stored in Celsius, humidity in percent, rainfall and evaporation in millimetres, and wind in km/h.
- SMOTE was applied separately to the crop and rice-pest class targets to reduce class imbalance.
